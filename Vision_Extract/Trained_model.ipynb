```python
import os, sys, subprocess, random, time
from pathlib import Path
import numpy as np
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt
import cv2
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.optim.lr_scheduler import ReduceLROnPlateau, CosineAnnealingLR, StepLR
import torchvision.models.segmentation as segmentation_models

import albumentations as A
from albumentations.pytorch import ToTensorV2
from scipy import ndimage as ndi

CONFIG = {
    "images_dir": "/kaggle/input/coco-2017-dataset/coco2017/val2017",
    "masks_dir": "/kaggle/working/val2017_masks",
    "num_epochs": 30,          
    "batch_size": 8,
    "num_workers": 2,
    "img_size": (512,512),
    "model_name": "deeplabv3_resnet50",
    "num_classes": 2,
    "optimizer_name": "Adam",
    "lr": 1e-4,
    "weight_decay": 1e-5,
    "scheduler": "plateau",
    "step_step_size": 10,
    "step_gamma": 0.5,
    "cosine_T_max": 40,
    "plateau_patience": 3,
    "device": "cuda" if torch.cuda.is_available() else "cpu",
    "mixed_precision": True,
    "early_stopping_patience": 6,
    "save_path": "/kaggle/working/best_seg_model.pth",
    "seed": 42,
    "val_fraction": 0.1,
    "tta_flip": True,
    "use_crf": True,            
    "crf_iters": 5,
    "fg_thresh": 0.5,
    "min_area": 300,
    "keep_largest": False,
    "pred_masks_dir": "/kaggle/working/pred_masks",
    "preview_samples": 15,  }    

random.seed(CONFIG["seed"])
np.random.seed(CONFIG["seed"])
torch.manual_seed(CONFIG["seed"])
Path(CONFIG["pred_masks_dir"]).mkdir(parents=True, exist_ok=True)

train_transform = A.Compose([
    A.RandomResizedCrop(size=CONFIG["img_size"], scale=(0.8, 1.0), ratio=(0.75,1.33), p=1.0),
    A.HorizontalFlip(p=0.5),
    A.Rotate(limit=15, p=0.4),
    A.RandomBrightnessContrast(p=0.4),
    A.OneOf([A.MotionBlur(p=0.3), A.GaussNoise(p=0.2)], p=0.3),
    A.Normalize(),
    ToTensorV2()
], additional_targets={"mask": "mask"})

val_transform = A.Compose([
    A.Resize(height=CONFIG["img_size"][0], width=CONFIG["img_size"][1]),
    A.Normalize(),
    ToTensorV2()
], additional_targets={"mask": "mask"})

class SegmentationDataset(Dataset):
    def __init__(self, image_paths, masks_dir, transforms=None):
        self.image_paths = image_paths
        self.masks_dir = Path(masks_dir)
        self.transforms = transforms
    def __len__(self):
        return len(self.image_paths)
    def __getitem__(self, idx):
        p = self.image_paths[idx]
        img = np.array(Image.open(p).convert("RGB"))
        mask_path = self.masks_dir / (p.stem + "_mask.png")
        if mask_path.exists():
            mask = np.array(Image.open(mask_path).convert("L"))
        else:
            mask = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)
        mask = (mask > 127).astype(np.uint8)
        if self.transforms:
            aug = self.transforms(image=img, mask=mask)
            img_t = aug["image"]
            mask_t = aug["mask"]
            if isinstance(mask_t, torch.Tensor) and mask_t.dim() == 3 and mask_t.shape[0] == 1:
                mask_t = mask_t.squeeze(0)
            return img_t, mask_t.long()
        else:
            h,w = CONFIG["img_size"]
            img_r = cv2.resize(img, (w,h)).astype(np.float32)/255.0
            mean = np.array([0.485,0.456,0.406]); std = np.array([0.229,0.224,0.225])
            img_r = (img_r - mean[None,None,:]) / std[None,None,:]
            img_r = img_r.transpose(2,0,1)
            img_t = torch.from_numpy(img_r).float()
            mask_r = Image.fromarray((mask*255).astype(np.uint8)).resize((w,h), resample=Image.NEAREST)
            mask_t = torch.from_numpy(np.array(mask_r)//255).long()
            return img_t, mask_t

img_dir = Path(CONFIG["images_dir"])
mask_dir = Path(CONFIG["masks_dir"])
all_imgs = sorted([p for p in img_dir.iterdir() if p.suffix.lower() in ('.jpg','.jpeg','.png')])
all_imgs = [p for p in all_imgs if (mask_dir / (p.stem + "_mask.png")).exists()]
print(f"Found {len(all_imgs)} image-mask pairs (images_dir={img_dir}, masks_dir={mask_dir})")

random.shuffle(all_imgs)
num_total = len(all_imgs)
num_val = max(1, int(num_total * CONFIG["val_fraction"])) if num_total>0 else 0
val_list = all_imgs[:num_val]
train_list = all_imgs[num_val:]

print(f"Train: {len(train_list)} | Val: {len(val_list)}")

train_ds = SegmentationDataset(train_list, CONFIG["masks_dir"], transforms=train_transform)
val_ds   = SegmentationDataset(val_list, CONFIG["masks_dir"], transforms=val_transform)

train_loader = DataLoader(train_ds, batch_size=CONFIG["batch_size"], shuffle=True, num_workers=CONFIG["num_workers"], pin_memory=True)
val_loader   = DataLoader(val_ds, batch_size=CONFIG["batch_size"], shuffle=False, num_workers=CONFIG["num_workers"], pin_memory=True)

def get_model(name="deeplabv3_resnet50", num_classes=2):
    if name == "deeplabv3_resnet50":
        model = segmentation_models.deeplabv3_resnet50(pretrained=False, progress=True, num_classes=num_classes)
    elif name == "fcn_resnet50":
        model = segmentation_models.fcn_resnet50(pretrained=False, progress=True, num_classes=num_classes)
    else:
        raise ValueError("Unknown model")
    return model

device = torch.device(CONFIG["device"])
model = get_model(CONFIG["model_name"], num_classes=CONFIG["num_classes"]).to(device)
print("Using device:", device)
ce_loss = nn.CrossEntropyLoss()
def dice_loss_logits(pred, target, eps=1e-6):
    num_classes = pred.shape[1]
    pred_soft = torch.softmax(pred, dim=1)
    target_onehot = torch.nn.functional.one_hot(target, num_classes=num_classes).permute(0,3,1,2).float()
    dice = 0.0
    for c in range(num_classes):
        p = pred_soft[:,c,:,:]; t = target_onehot[:,c,:,:]
        inter = (p * t).sum(dim=(1,2)); denom = p.sum(dim=(1,2)) + t.sum(dim=(1,2)) + eps
        dice += (2.0 * inter / denom).mean()
    return 1.0 - dice / num_classes

def combo_loss_fn(logits, targets, alpha=0.5):
    return alpha * ce_loss(logits, targets) + (1.0 - alpha) * dice_loss_logits(logits, targets)

opt_name = CONFIG["optimizer_name"].lower()
if opt_name == "adam":
    optimizer = optim.Adam(model.parameters(), lr=CONFIG["lr"], weight_decay=CONFIG["weight_decay"])
elif opt_name == "adamw":
    optimizer = optim.AdamW(model.parameters(), lr=CONFIG["lr"], weight_decay=CONFIG["weight_decay"])
elif opt_name == "sgd":
    optimizer = optim.SGD(model.parameters(), lr=CONFIG["lr"], momentum=0.9, weight_decay=CONFIG["weight_decay"])
else:
    raise ValueError("Unknown optimizer")

sch_name = CONFIG["scheduler"].lower()
if sch_name == "plateau":
    scheduler = ReduceLROnPlateau(optimizer, mode='max', patience=CONFIG["plateau_patience"], factor=0.5, verbose=True)
elif sch_name == "cosine":
    scheduler = CosineAnnealingLR(optimizer, T_max=CONFIG["cosine_T_max"])
elif sch_name == "step":
    scheduler = StepLR(optimizer, step_size=CONFIG["step_step_size"], gamma=CONFIG["step_gamma"])
else:
    scheduler = None

def reset_stats(num_classes):
    return {"TP": np.zeros(num_classes, dtype=np.float64), "FP": np.zeros(num_classes, dtype=np.float64), "FN": np.zeros(num_classes, dtype=np.float64), "TN": np.zeros(num_classes, dtype=np.float64), "total_pixels": 0, "correct_pixels": 0}

def accumulate_batch_stats(preds, targets, stats):
    num_classes = len(stats["TP"])
    for c in range(num_classes):
        pred_c = (preds == c); targ_c = (targets == c)
        stats["TP"][c] += np.logical_and(pred_c, targ_c).sum()
        stats["FP"][c] += np.logical_and(pred_c, np.logical_not(targ_c)).sum()
        stats["FN"][c] += np.logical_and(np.logical_not(pred_c), targ_c).sum()
        stats["TN"][c] += np.logical_and(np.logical_not(pred_c), np.logical_not(targ_c)).sum()
    stats["total_pixels"] += preds.size
    stats["correct_pixels"] += (preds == targets).sum()

def compute_final_metrics(stats):
    eps = 1e-8
    num_classes = len(stats["TP"])
    ious = []; dices = []
    for c in range(num_classes):
        tp, fp, fn = stats["TP"][c], stats["FP"][c], stats["FN"][c]
        denom = tp + fp + fn
        if denom > 0:
            ious.append(tp / (denom + eps))
            dices.append((2*tp) / (2*tp + fp + fn + eps))
    mean_iou = float(np.mean(ious)) if len(ious)>0 else 0.0
    mean_dice = float(np.mean(dices)) if len(dices)>0 else 0.0
    pixel_acc = float(stats["correct_pixels"] / (stats["total_pixels"] + eps))
    return pixel_acc, mean_iou, mean_dice, ious, dices

scaler = torch.cuda.amp.GradScaler(enabled=CONFIG["mixed_precision"])
best_miou = 0.0; patience = 0; start_time = time.time()

for epoch in range(1, CONFIG["num_epochs"] + 1):
    model.train()
    train_loss_accum = 0.0
    for imgs, masks in tqdm(train_loader, desc=f"Epoch {epoch} Train", leave=False):
        imgs = imgs.to(device, non_blocking=True); masks = masks.to(device, non_blocking=True)
        optimizer.zero_grad()
        with torch.cuda.amp.autocast(enabled=CONFIG["mixed_precision"]):
            logits = model(imgs)['out']
            loss = combo_loss_fn(logits, masks, alpha=0.5)
        scaler.scale(loss).backward()
        scaler.step(optimizer); scaler.update()
        train_loss_accum += loss.item() * imgs.size(0)
    train_loss = train_loss_accum / max(1,len(train_ds))

    model.eval()
    stats = reset_stats(CONFIG["num_classes"])
    val_loss_accum = 0.0
    with torch.no_grad():
        for imgs, masks in tqdm(val_loader, desc=f"Epoch {epoch} Val", leave=False):
            imgs = imgs.to(device, non_blocking=True); masks = masks.to(device, non_blocking=True)
            with torch.cuda.amp.autocast(enabled=CONFIG["mixed_precision"]):
                logits = model(imgs)['out']
                loss = combo_loss_fn(logits, masks, alpha=0.5)
            val_loss_accum += loss.item() * imgs.size(0)
            preds = torch.argmax(logits, dim=1).cpu().numpy()
            targs = masks.cpu().numpy()
            accumulate_batch_stats(preds, targs, stats)
    val_loss = val_loss_accum / max(1,len(val_ds))
    pixel_acc, mean_iou, mean_dice, ious_per_class, dices_per_class = compute_final_metrics(stats)


    if CONFIG["scheduler"].lower() == "plateau" and scheduler is not None:
        scheduler.step(mean_iou)
    elif scheduler is not None and CONFIG["scheduler"].lower() != "plateau":
        scheduler.step()

    print(f"Epoch {epoch} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | Pixel Acc: {pixel_acc:.4f} | Mean IoU: {mean_iou:.4f} | Mean Dice: {mean_dice:.4f}")


    if mean_iou > best_miou + 1e-6:
        best_miou = mean_iou; patience = 0
        torch.save({"epoch": epoch, "model_state": model.state_dict(), "optimizer_state": optimizer.state_dict(), "mean_iou": mean_iou}, CONFIG["save_path"])
        print(f"Saved best model (mIoU={best_miou:.4f}) -> {CONFIG['save_path']}")
    else:
        patience += 1
        if patience >= CONFIG["early_stopping_patience"]:
            print("Early stopping triggered.")
            break

elapsed = time.time() - start_time
print(f"Training finished in {elapsed/60:.2f} minutes. Best val mIoU: {best_miou:.4f}")


if CONFIG["use_crf"]:
    try:
        import pydensecrf.densecrf as dcrf
        from pydensecrf.utils import unary_from_softmax
    except Exception as e:
        print("Unexpected CRF import error, disabling CRF:", e)
        CONFIG["use_crf"] = False

def preprocess_image_cv2(img_rgb, size):
    h,w = size
    img = cv2.resize(img_rgb, (w,h), interpolation=cv2.INTER_LINEAR).astype(np.float32)/255.0
    mean = np.array([0.485,0.456,0.406], dtype=np.float32); std = np.array([0.229,0.224,0.225], dtype=np.float32)
    img = (img - mean[None,None,:]) / std[None,None,:]
    img_chw = img.transpose(2,0,1)
    tensor = torch.from_numpy(img_chw).unsqueeze(0).float()
    return tensor

def crf_refine(prob_fg, img_rgb, n_iters=5):
    if not CONFIG["use_crf"]:
        return prob_fg
    h,w = prob_fg.shape
    probs = np.stack([1.0 - prob_fg, prob_fg], axis=0).astype(np.float32)
    d = dcrf.DenseCRF2D(w, h, 2)
    unary = unary_from_softmax(probs)
    d.setUnaryEnergy(unary)
    d.addPairwiseGaussian(sxy=3, compat=3)
    d.addPairwiseBilateral(sxy=60, srgb=5, rgbim=img_rgb, compat=5)
    Q = d.inference(n_iters)
    res = np.array(Q).reshape((2,h,w))[1]
    return np.clip(res, 0.0, 1.0).astype(np.float32)

def postprocess_mask(prob_map, fg_thresh=0.5, min_area=300, keep_largest=False):
    mask = (prob_map >= fg_thresh).astype(np.uint8)
    kernel_close = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (11,11))
    kernel_open  = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel_close)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel_open)
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    out = np.zeros_like(mask); comps=[]
    for lab in range(1, num_labels):
        area = stats[lab, cv2.CC_STAT_AREA]
        if area >= min_area:
            comps.append((lab, area)); out[labels==lab]=1
    if keep_largest and len(comps)>0:
        comps.sort(key=lambda x: x[1], reverse=True); largest_lab = comps[0][0]; out = (labels==largest_lab).astype(np.uint8)
    out = ndi.binary_fill_holes(out).astype(np.uint8)
    out = cv2.morphologyEx(out.astype(np.uint8), cv2.MORPH_OPEN, kernel_open)
    return (out * 255).astype(np.uint8)

def predict_with_tta(img_rgb, model, device, size, tta_flip=True, use_crf=False):
    inp = preprocess_image_cv2(img_rgb, size).to(device)
    with torch.no_grad():
        out = model(inp)['out']; prob = F.softmax(out, dim=1).cpu().numpy()[0,1,:,:]
    probs = [prob]
    if tta_flip:
        img_f = img_rgb[:, ::-1, :].copy()
        inp_f = preprocess_image_cv2(img_f, size).to(device)
        with torch.no_grad():
            out_f = model(inp_f)['out']; prob_f = F.softmax(out_f, dim=1).cpu().numpy()[0,1,:,:]
        prob_f = prob_f[:, ::-1]; probs.append(prob_f)
    avg_prob = np.mean(np.stack(probs, axis=0), axis=0)
    if use_crf:
        img_resized = cv2.resize(img_rgb, (size[1], size[0]), interpolation=cv2.INTER_LINEAR)
        avg_prob = crf_refine(avg_prob.astype(np.float32), img_resized, n_iters=CONFIG["crf_iters"])
    return avg_prob.astype(np.float32)


img_files = sorted([p for p in img_dir.iterdir() if p.suffix.lower() in ('.jpg','.jpeg','.png')])
img_files = [p for p in img_files if (mask_dir / (p.stem + "_mask.png")).exists()]
print("Predicting on", len(img_files), "images...")

stats = {"TP": np.zeros(2, dtype=np.float64), "FP": np.zeros(2, dtype=np.float64), "FN": np.zeros(2, dtype=np.float64), "TN": np.zeros(2, dtype=np.float64), "total_pixels": 0, "correct_pixels": 0}

for p in tqdm(img_files, desc="Predicting"):
    try:
        img = np.array(Image.open(p).convert("RGB"))
    except Exception as e:
        print("skip", p, e); continue
    prob_map = predict_with_tta(img, model, device, CONFIG["img_size"], tta_flip=CONFIG["tta_flip"], use_crf=CONFIG["use_crf"])
    pred_mask = postprocess_mask(prob_map, fg_thresh=CONFIG["fg_thresh"], min_area=CONFIG["min_area"], keep_largest=CONFIG["keep_largest"])
    gt_path = mask_dir / (p.stem + "_mask.png")
    gt_mask_full = np.array(Image.open(gt_path).convert("L"))
    gt_mask_resized = cv2.resize(gt_mask_full, (CONFIG["img_size"][1], CONFIG["img_size"][0]), interpolation=cv2.INTER_NEAREST)
    gt_mask_bin = (gt_mask_resized > 127).astype(np.uint8) * 255

    pred_bin = (pred_mask > 127).astype(np.uint8); gt_bin = (gt_mask_bin > 127).astype(np.uint8)
    tp = np.logical_and(pred_bin==1, gt_bin==1).sum(); fp = np.logical_and(pred_bin==1, gt_bin==0).sum()
    fn = np.logical_and(pred_bin==0, gt_bin==1).sum(); tn = np.logical_and(pred_bin==0, gt_bin==0).sum()
    stats["TP"][1] += tp; stats["FP"][1] += fp; stats["FN"][1] += fn; stats["TN"][1] += tn
    stats["TP"][0] += tn; stats["FP"][0] += fn; stats["FN"][0] += fp; stats["TN"][0] += tp
    stats["correct_pixels"] += (pred_bin == gt_bin).sum(); stats["total_pixels"] += pred_bin.size
    out_path = Path(CONFIG["pred_masks_dir"]) / (p.stem + "_pred.png"); cv2.imwrite(str(out_path), pred_mask)


eps = 1e-8; ious=[]; dices=[]
for c in range(2):
    tp = stats["TP"][c]; fp = stats["FP"][c]; fn = stats["FN"][c]
    denom = tp + fp + fn
    if denom > 0:
        ious.append(tp / (denom + eps)); dices.append((2*tp)/(2*tp + fp + fn + eps))
mean_iou = float(np.mean(ious)) if len(ious)>0 else 0.0
mean_dice = float(np.mean(dices)) if len(dices)>0 else 0.0
pixel_acc = float(stats["correct_pixels"] / (stats["total_pixels"] + eps))


print(f"Overall Pixel Accuracy: {pixel_acc:.4f}")
print(f"Mean IoU: {mean_iou:.4f}  (per-class IoUs: {['{:.4f}'.format(x) for x in ious]})")
print(f"Mean Dice: {mean_dice:.4f} (per-class Dice: {['{:.4f}'.format(x) for x in dices]})")

pred_dir = Path(CONFIG["pred_masks_dir"])
img_files = sorted([p for p in img_dir.iterdir() if p.suffix.lower() in ('.jpg','.jpeg','.png')])
img_files = [p for p in img_files if (mask_dir / (p.stem + "_mask.png")).exists() and (pred_dir / (p.stem + "_pred.png")).exists()]

def compute_per_image_metrics(img_paths, mask_dir, pred_dir):
    records=[]; eps=1e-8
    for p in img_paths:
        gt_path = mask_dir / (p.stem + "_mask.png"); pred_path = pred_dir / (p.stem + "_pred.png")
        if not gt_path.exists() or not pred_path.exists(): continue
        gt = cv2.imread(str(gt_path), cv2.IMREAD_GRAYSCALE); pred = cv2.imread(str(pred_path), cv2.IMREAD_GRAYSCALE)
        if gt.shape != pred.shape:
            gt = cv2.resize(gt, (pred.shape[1], pred.shape[0]), interpolation=cv2.INTER_NEAREST)
        gt_bin = (gt>127).astype(np.uint8); pred_bin = (pred>127).astype(np.uint8)
        tp = int(np.logical_and(pred_bin==1, gt_bin==1).sum()); fp = int(np.logical_and(pred_bin==1, gt_bin==0).sum())
        fn = int(np.logical_and(pred_bin==0, gt_bin==1).sum()); tn = int(np.logical_and(pred_bin==0, gt_bin==0).sum())
        total = gt_bin.size; pixel_acc = float((pred_bin == gt_bin).sum())/float(total+eps)
        iou = float(tp)/float(tp+fp+fn+eps) if (tp+fp+fn)>0 else 0.0
        dice = float(2*tp)/float(2*tp+fp+fn+eps) if (2*tp+fp+fn)>0 else 0.0
        fg_area = int(gt_bin.sum())
        records.append({"file_name":p.name,"stem":p.stem,"tp":tp,"fp":fp,"fn":fn,"tn":tn,"pixel_acc":pixel_acc,"iou":iou,"dice":dice,"fg_area":fg_area,"img_path":str(p),"gt_path":str(gt_path),"pred_path":str(pred_path)})
    df=pd.DataFrame.from_records(records); df.sort_values(by="file_name", inplace=True); return df

metrics_df = compute_per_image_metrics(img_files, mask_dir, pred_dir)
metrics_csv_path = Path(CONFIG["pred_masks_dir"]) / "per_image_metrics.csv"
metrics_df.to_csv(metrics_csv_path, index=False)
print(f"Per-image metrics saved to: {metrics_csv_path}")

def select_samples(mode="fixed_seed", n=13, seed=42, file_list=None):
    if len(metrics_df)==0: raise ValueError("metrics_df empty — ensure preds exist")
    if mode=="random": picks = random.sample(list(metrics_df['img_path']), min(n,len(metrics_df))); return [Path(p) for p in picks]
    if mode=="fixed_seed":
        rng = random.Random(seed); picks = rng.sample(list(metrics_df['img_path']), min(n,len(metrics_df))); return [Path(p) for p in picks]
    if mode=="first_n": picks = metrics_df['img_path'].tolist()[:n]; return [Path(p) for p in picks]
    if mode in ("largest_fg","smallest_fg"):
        df2 = metrics_df.copy(); df2 = df2.sort_values(by="fg_area", ascending=(mode=="smallest_fg")); picks = df2['img_path'].tolist()[:n]; return [Path(p) for p in picks]
    if mode in ("worst_iou","best_iou"):
        df2 = metrics_df.copy(); df2 = df2.sort_values(by="iou", ascending=(mode=="worst_iou")); picks = df2['img_path'].tolist()[:n]; return [Path(p) for p in picks]
    if mode=="list":
        if file_list is None: raise ValueError("file_list required for mode='list'")
        picks=[]
        for fname in file_list:
            if Path(fname).exists(): picks.append(str(Path(fname))); continue
            row = metrics_df[metrics_df['file_name']==fname]
            if len(row)==1: picks.append(row.iloc[0]['img_path']); continue
            row = metrics_df[metrics_df['stem']==fname]
            if len(row)==1: picks.append(row.iloc[0]['img_path']); continue
            print(f"Warning: {fname} not found; skipping.")
        return [Path(p) for p in picks][:n]
    raise ValueError("Unknown mode")

def preview_selected_samples(sample_paths):
    sample_paths=[Path(p) for p in sample_paths]; n=len(sample_paths)
    fig, axes = plt.subplots(n, 3, figsize=(12, 4*n))
    if n==1: axes = np.expand_dims(axes,0)
    for i,p in enumerate(sample_paths):
        orig = np.array(Image.open(p).convert("RGB").resize((CONFIG["img_size"][1], CONFIG["img_size"][0])))
        gt = cv2.resize(np.array(Image.open(mask_dir / (p.stem + "_mask.png")).convert("L")), (CONFIG["img_size"][1], CONFIG["img_size"][0]), interpolation=cv2.INTER_NEAREST)
        pred = cv2.imread(str(pred_dir / (p.stem + "_pred.png")), cv2.IMREAD_GRAYSCALE)
        axes[i,0].imshow(orig); axes[i,0].axis("off"); axes[i,0].set_title(f"Original\n{p.name}")
        axes[i,1].imshow(gt, cmap="gray"); axes[i,1].axis("off"); axes[i,1].set_title("Ground-truth")
        axes[i,2].imshow(pred, cmap="gray"); axes[i,2].axis("off"); axes[i,2].set_title("Predicted by model")
    plt.tight_layout(); plt.show()

n_show = min(CONFIG["preview_samples"], len(metrics_df))
if n_show > 0:
    samples = select_samples(mode="fixed_seed", n=n_show, seed=123)
    preview_selected_samples(samples)
else:
    print("No samples to preview (metrics_df empty).")


```
