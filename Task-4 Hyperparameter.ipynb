```python
# improved_train_and_eval_amp_fixed.py
import os, random, math, time
import numpy as np
from PIL import Image
import cv2
import matplotlib.pyplot as plt
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, random_split

import albumentations as A
from albumentations.pytorch import ToTensorV2
from pycocotools.coco import COCO

from torchvision.models.segmentation import (
    deeplabv3_resnet50, DeepLabV3_ResNet50_Weights,
    deeplabv3_resnet101, DeepLabV3_ResNet101_Weights
)

# ----------------------
# Config
# ----------------------
DATA_DIR = r"C:/Users/rahul/Downloads/coco"
ANNOTATIONS = os.path.join(DATA_DIR, "annotations", "instances_val2017.json")
IMAGES_DIR = os.path.join(DATA_DIR, "val2017")

SAVE_PATH = r"C:\Users\rahul\Downloads\infosys project\deeplabv3_best_safe.pth"
USE_RESNET101 = False   # set True to try ResNet101 (more memory)
INPUT_SIZE = 384        # 256 -> fast, 384 -> better edges, 512 -> best but needs memory
BATCH_SIZE = 2          # set 2 (reduce to 1 if OOM)
ACCUM_ITERS = 1         # set >1 to simulate larger batch via accumulation (if GPU mem low)
NUM_EPOCHS = 40
NUM_WORKERS = 0         # on Windows use 0
PATIENCE = 7            # early stopping patience on val IoU
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
USE_PIN_MEMORY = True if DEVICE.type == "cuda" else False
print("Device:", DEVICE)


# ----------------------
# Dataset
# ----------------------
class COCODataset(Dataset):
    def __init__(self, root, annFile, transforms=None):
        self.root = root
        self.coco = COCO(annFile)
        self.ids = list(self.coco.imgs.keys())
        self.transforms = transforms

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, idx):
        img_id = self.ids[idx]
        info = self.coco.loadImgs(img_id)[0]
        img_path = os.path.join(self.root, info['file_name'])
        img = np.array(Image.open(img_path).convert("RGB"))

        # build binary mask combining all object masks
        mask = np.zeros((info['height'], info['width']), dtype=np.uint8)
        ann_ids = self.coco.getAnnIds(imgIds=img_id)
        anns = self.coco.loadAnns(ann_ids)
        for ann in anns:
            mask = np.maximum(mask, self.coco.annToMask(ann).astype(np.uint8))

        if self.transforms:
            augmented = self.transforms(image=img, mask=mask)
            img = augmented['image']
            mask = augmented['mask']

        # img: tensor CxHxW, mask: HxW (Torch Tensor from ToTensorV2 may be uint8)
        return img, mask.long()

# ----------------------
# Transforms (lighter)
# ----------------------
train_transforms = A.Compose([
    A.Resize(INPUT_SIZE, INPUT_SIZE),
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.2),
    A.Normalize(mean=(0.485,0.456,0.406), std=(0.229,0.224,0.225)),
    ToTensorV2()
])

val_transforms = A.Compose([
    A.Resize(INPUT_SIZE, INPUT_SIZE),
    A.Normalize(mean=(0.485,0.456,0.406), std=(0.229,0.224,0.225)),
    ToTensorV2()
])

# ----------------------
# Losses + metrics
# ----------------------
class DiceLoss(nn.Module):
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth
    def forward(self, logits, targets):
        probs = torch.sigmoid(logits)
        targets = targets.float()
        # flatten
        probs_flat = probs.view(probs.size(0), -1)
        targ_flat = targets.view(targets.size(0), -1)
        intersection = (probs_flat * targ_flat).sum(dim=1)
        dice = (2. * intersection + self.smooth) / (probs_flat.sum(dim=1) + targ_flat.sum(dim=1) + self.smooth)
        return 1.0 - dice.mean()

bce_loss = nn.BCEWithLogitsLoss()

def combined_loss(outputs, masks):
    return 0.5 * bce_loss(outputs, masks.float()) + 0.5 * DiceLoss()(outputs, masks)

def compute_batch_metrics(preds, masks, threshold=0.5):
    # preds: [B,1,H,W] probabilities
    preds_bin = (preds > threshold).float()
    masks_f = masks.float()
    # compute IoU/Dice/Acc per sample
    eps = 1e-6
    intersection = (preds_bin * masks_f).view(preds.size(0), -1).sum(dim=1)
    union = (preds_bin + masks_f).view(preds.size(0), -1).sum(dim=1) - intersection
    iou = (intersection + eps) / (union + eps)
    dice = (2*intersection + eps) / ((preds_bin.view(preds.size(0), -1).sum(dim=1) + masks_f.view(masks_f.size(0), -1).sum(dim=1)) + eps)
    acc = (preds_bin == masks_f).float().view(preds.size(0), -1).mean(dim=1)
    return iou.mean().item(), dice.mean().item(), acc.mean().item()

# ----------------------
# Post-processing (inference)
# ----------------------
def postprocess_mask(mask, min_area=1500):
    # mask: np.uint8 0/1
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))
    m = cv2.morphologyEx(mask.astype('uint8'), cv2.MORPH_CLOSE, kernel, iterations=2)
    m = cv2.morphologyEx(m, cv2.MORPH_OPEN, kernel, iterations=1)
    # remove small components
    num, labels, stats, _ = cv2.connectedComponentsWithStats(m, connectivity=8)
    out = np.zeros_like(m)
    for i in range(1, num):
        if stats[i, cv2.CC_STAT_AREA] >= min_area:
            out[labels == i] = 1
    return out

# ----------------------
# Train & Validate
# ----------------------
def train_model(model, train_loader, val_loader, device, save_path, epochs=NUM_EPOCHS):
    optimizer = optim.Adam(model.parameters(), lr=1e-4, weight_decay=1e-5)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)
    best_iou = 0.0
    patience_counter = 0

    use_amp = (device.type == "cuda")
    if use_amp:
        scaler = torch.amp.GradScaler("cuda")

    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        pbar = tqdm(enumerate(train_loader), total=len(train_loader), desc=f"Epoch {epoch+1}/{epochs}")
        optimizer.zero_grad()
        for step, (imgs, masks) in pbar:
            imgs = imgs.to(device, non_blocking=True)
            masks = masks.to(device, non_blocking=True).unsqueeze(1)  # [B,1,H,W]
            if use_amp:
                with torch.amp.autocast("cuda"):
                    outputs = model(imgs)['out']
                    loss = 0.5 * bce_loss(outputs, masks.float()) + 0.5 * DiceLoss()(outputs, masks)
                scaler.scale(loss).backward()
                if (step + 1) % ACCUM_ITERS == 0:
                    scaler.step(optimizer)
                    scaler.update()
                    optimizer.zero_grad()
            else:
                outputs = model(imgs)['out']
                loss = 0.5 * bce_loss(outputs, masks.float()) + 0.5 * DiceLoss()(outputs, masks)
                loss.backward()
                if (step + 1) % ACCUM_ITERS == 0:
                    optimizer.step()
                    optimizer.zero_grad()

            running_loss += loss.item()
            pbar.set_postfix({"loss": loss.item()})

        # Validation
        model.eval()
        val_loss = 0.0
        val_iou = 0.0
        val_dice = 0.0
        val_acc = 0.0
        with torch.no_grad():
            for imgs, masks in val_loader:
                imgs = imgs.to(device)
                masks = masks.to(device).unsqueeze(1)
                if use_amp:
                    with torch.amp.autocast("cuda"):
                        outputs = model(imgs)['out']
                        loss = 0.5 * bce_loss(outputs, masks.float()) + 0.5 * DiceLoss()(outputs, masks)
                else:
                    outputs = model(imgs)['out']
                    loss = 0.5 * bce_loss(outputs, masks.float()) + 0.5 * DiceLoss()(outputs, masks)
                val_loss += loss.item()
                probs = torch.sigmoid(outputs)
                iou, dice, acc = compute_batch_metrics(probs, masks)
                val_iou += iou
                val_dice += dice
                val_acc += acc

        val_loss /= len(val_loader)
        val_iou /= len(val_loader)
        val_dice /= len(val_loader)
        val_acc /= len(val_loader)
        print(f"\nEpoch {epoch+1} summary: TrainLoss={running_loss/len(train_loader):.4f} | ValLoss={val_loss:.4f} | IoU={val_iou:.4f} | Dice={val_dice:.4f} | Acc={val_acc:.4f}")

        scheduler.step(val_loss)

        # save best
        if val_iou > best_iou + 1e-4:
            best_iou = val_iou
            print(f"New best IoU {best_iou:.4f} -> saving model to {save_path}")
            try:
                torch.save(model.cpu().state_dict(), save_path)   # save on CPU
                model.to(device)
            except Exception as e:
                print("Error saving model:", e)
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= PATIENCE:
                print("Early stopping triggered.")
                break

    print("Training finished. Best IoU:", best_iou)
    return best_iou

# ----------------------
# Utility: visualize predictions from val set (TTA + postprocess)
# ----------------------
def visualize_random_predictions(model, coco_ann, img_root, device, num=12, tta=True, threshold=0.5):
    model.eval()
    img_ids = random.sample(coco_ann.getImgIds(), num)
    for img_id in img_ids:
        info = coco_ann.loadImgs(img_id)[0]
        path = os.path.join(img_root, info['file_name'])
        img = np.array(Image.open(path).convert("RGB"))
        # build GT mask
        mask = np.zeros((info['height'], info['width']), dtype=np.uint8)
        for ann in coco_ann.loadAnns(coco_ann.getAnnIds(imgIds=img_id)):
            mask = np.maximum(mask, coco_ann.annToMask(ann).astype(np.uint8))
        # resize for comparison
        resized_gt = A.Resize(INPUT_SIZE, INPUT_SIZE)(image=mask)['image']

        # prepare input
        aug = val_transforms(image=img)
        inp = aug['image'].unsqueeze(0).to(device)

        with torch.no_grad():
            if tta:
                # original
                p0 = torch.sigmoid(model(inp)['out']).cpu().numpy()[0,0]
                # flip
                img_h = np.fliplr(img).copy()
                p1 = torch.sigmoid(model(val_transforms(image=img_h)['image'].unsqueeze(0).to(device))['out']).cpu().numpy()[0,0]
                p1 = np.fliplr(p1)
                prob_map = (p0 + p1) / 2.0
            else:
                prob_map = torch.sigmoid(model(inp)['out']).cpu().numpy()[0,0]

            pred = (prob_map > threshold).astype(np.uint8)
            pred_pp = postprocess_mask(pred, min_area=800)

        # show
        plt.figure(figsize=(12,4))
        plt.subplot(1,3,1); plt.imshow(img); plt.title("Image"); plt.axis('off')
        plt.subplot(1,3,2); plt.imshow(resized_gt, cmap='gray'); plt.title("Ground Truth"); plt.axis('off')
        plt.subplot(1,3,3); plt.imshow(pred_pp, cmap='gray'); plt.title("Predicted (post-processed)"); plt.axis('off')
        plt.show()


# ----------------------
# Main
# ----------------------
if __name__ == "__main__":
    # Prepare dataset / dataloaders
    full_dataset = COCODataset(IMAGES_DIR, ANNOTATIONS, transforms=train_transforms)
    n = len(full_dataset)
    print("Dataset size:", n)
    train_n = int(0.8 * n)
    val_n = n - train_n
    train_ds, val_ds = random_split(full_dataset, [train_n, val_n])
    # ensure val uses val_transforms
    val_ds.dataset.transforms = val_transforms

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=USE_PIN_MEMORY)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=USE_PIN_MEMORY)

    # build model
    if USE_RESNET101:
        weights = DeepLabV3_ResNet101_Weights.DEFAULT
        model = deeplabv3_resnet101(weights=weights)
    else:
        weights = DeepLabV3_ResNet50_Weights.DEFAULT
        model = deeplabv3_resnet50(weights=weights)

    # replace final classifier for binary segmentation
    model.classifier[4] = nn.Conv2d(256, 1, kernel_size=1)
    model.to(DEVICE)

    # train
    best = train_model(model, train_loader, val_loader, DEVICE, SAVE_PATH, epochs=NUM_EPOCHS)

    # load best and visualize a few predictions
    try:
        st = torch.load(SAVE_PATH, map_location=DEVICE)
        model.load_state_dict(st, strict=False)
        model.to(DEVICE)
    except Exception as e:
        print("Could not load saved model:", e)

    coco = COCO(ANNOTATIONS)
    visualize_random_predictions(model, coco, IMAGES_DIR, DEVICE, num=12, tta=True, threshold=0.5)
```
