```python
import os
import json
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import albumentations as A
from albumentations.pytorch import ToTensorV2
from torchvision.models.segmentation import deeplabv3_resnet50, DeepLabV3_ResNet50_Weights
import cv2
import matplotlib.pyplot as plt
from tqdm import tqdm
import io
from typing import Tuple, List, Dict, Optional

# ==================== Device Setup ====================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ==================== Model Configuration ====================
NUM_CLASSES = 2
IMAGE_SIZE = 256

# ==================== Segmentation Model Class ====================
class SegmentationModel:
    def __init__(self, model_path: str = None):
        self.device = device
        self.model = None
        self.transform = self._setup_transforms()

        if model_path and os.path.exists(model_path):
            self.load_model(model_path)
        else:
            self._initialize_model()

    def _setup_transforms(self) -> A.Compose:
        """Setup image transformations for inference"""
        return A.Compose([
            A.Resize(IMAGE_SIZE, IMAGE_SIZE),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ])

    def _initialize_model(self):
        """Initialize model architecture"""
        weights = DeepLabV3_ResNet50_Weights.DEFAULT
        self.model = deeplabv3_resnet50(weights=weights, aux_loss=True)

        # Modify classifier for binary segmentation
        old_cls = self.model.classifier
        self.model.classifier = nn.Sequential(
            old_cls[0],
            old_cls[1],
            old_cls[2],
            nn.Dropout(0.3),
            nn.Conv2d(256, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.Conv2d(128, NUM_CLASSES, kernel_size=1)
        )
        self.model = self.model.to(self.device)
        print("✓ Model initialized with pretrained weights")

    def load_model(self, model_path: str):
        """Load trained model weights"""
        if self.model is None:
            self._initialize_model()

        try:
            state_dict = torch.load(model_path, map_location=self.device)
            self.model.load_state_dict(state_dict)
            self.model.eval()
            print(f"✓ Model loaded from: {model_path}")
        except Exception as e:
            print(f"❌ Error loading model: {e}")
            print("⚠️ Using initialized model instead")

    def preprocess_image(self, image_data: bytes) -> np.ndarray:
        """Convert image bytes to numpy array"""
        image = Image.open(io.BytesIO(image_data))
        return np.array(image.convert("RGB"))

    def predict_mask(self, image: np.ndarray, use_tta: bool = True) -> np.ndarray:
        """Generate binary mask with optional Test Time Augmentation"""
        original_h, original_w = image.shape[:2]

        # Apply transformations
        transformed = self.transform(image=image)
        img_tensor = transformed["image"].unsqueeze(0).to(self.device)

        with torch.no_grad():
            if use_tta:
                # Test Time Augmentation for better accuracy
                outputs = []

                # Original
                outputs.append(self.model(img_tensor)['out'])

                # Horizontal flip
                flipped_h = torch.flip(img_tensor, [3])
                out_flipped_h = self.model(flipped_h)['out']
                outputs.append(torch.flip(out_flipped_h, [3]))

                # Vertical flip
                flipped_v = torch.flip(img_tensor, [2])
                out_flipped_v = self.model(flipped_v)['out']
                outputs.append(torch.flip(out_flipped_v, [2]))

                # Average predictions
                avg_output = torch.mean(torch.stack(outputs), dim=0)
                pred_mask = torch.argmax(avg_output, dim=1).squeeze(0).cpu().numpy()
            else:
                # Single prediction
                output = self.model(img_tensor)['out']
                pred_mask = torch.argmax(output, dim=1).squeeze(0).cpu().numpy()

        # Resize to original dimensions
        mask = cv2.resize(pred_mask.astype(np.uint8),
                         (original_w, original_h),
                         interpolation=cv2.INTER_LINEAR)
        return mask

    def postprocess_mask(self, mask: np.ndarray) -> np.ndarray:
        """Clean and refine predicted mask"""
        # Morphological operations
        kernel = np.ones((3, 3), np.uint8)
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
        mask = cv2.dilate(mask, kernel, iterations=1)

        # Keep largest connected component
        num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(mask, 8)
        if num_labels > 1:
            largest_label = np.argmax(stats[1:, cv2.CC_STAT_AREA]) + 1
            mask = (labels == largest_label).astype(np.uint8)

        # Smooth edges
        mask_float = mask.astype(np.float32)
        mask_smooth = cv2.GaussianBlur(mask_float, (5, 5), 0)
        mask = (mask_smooth > 0.5).astype(np.uint8)

        return mask

    def extract_object(self, image: np.ndarray, mask: np.ndarray,
                      background: Tuple[int, int, int] = (0, 0, 0)) -> np.ndarray:
        """Extract object with specified background"""
        # Create smooth alpha mask
        mask_float = mask.astype(np.float32)
        alpha_mask = cv2.GaussianBlur(mask_float, (7, 7), 0)
        alpha_mask = np.clip(alpha_mask, 0, 1)
        alpha_mask = np.expand_dims(alpha_mask, axis=-1)

        # Blend with background
        background_img = np.ones_like(image) * background
        result = image * alpha_mask + background_img * (1 - alpha_mask)

        return result.astype(np.uint8)

    def process_single_image(self, image_data: bytes,
                           background: Tuple[int, int, int] = (0, 0, 0),
                           use_tta: bool = True) -> Dict:
        """Complete processing pipeline for single image"""
        try:
            # Preprocess
            original_image = self.preprocess_image(image_data)

            # Predict mask
            raw_mask = self.predict_mask(original_image, use_tta=use_tta)

            # Postprocess mask
            clean_mask = self.postprocess_mask(raw_mask)

            # Extract object
            result_image = self.extract_object(original_image, clean_mask, background)

            # Calculate metrics
            foreground_ratio = np.sum(clean_mask) / clean_mask.size

            return {
                'success': True,
                'original_image': original_image,
                'result_image': result_image,
                'mask': clean_mask,
                'foreground_ratio': foreground_ratio,
                'original_shape': original_image.shape,
                'result_shape': result_image.shape
            }

        except Exception as e:
            return {
                'success': False,
                'error': str(e)
            }

    def process_batch(self, image_data_list: List[bytes],
                     background: Tuple[int, int, int] = (0, 0, 0)) -> List[Dict]:
        """Process multiple images"""
        results = []
        for i, image_data in enumerate(tqdm(image_data_list, desc="Processing batch")):
            result = self.process_single_image(image_data, background)
            result['index'] = i
            results.append(result)
        return results

# ==================== Model Evaluator ====================
class ModelEvaluator:
    def __init__(self, model: SegmentationModel):
        self.model = model

    def calculate_iou(self, pred_mask: np.ndarray, true_mask: np.ndarray) -> float:
        """Calculate Intersection over Union"""
        intersection = np.logical_and(pred_mask, true_mask).sum()
        union = np.logical_or(pred_mask, true_mask).sum()
        return intersection / union if union > 0 else 0.0

    def calculate_dice(self, pred_mask: np.ndarray, true_mask: np.ndarray) -> float:
        """Calculate Dice coefficient"""
        intersection = np.logical_and(pred_mask, true_mask).sum() * 2
        total = pred_mask.sum() + true_mask.sum()
        return intersection / total if total > 0 else 0.0

    def calculate_accuracy(self, pred_mask: np.ndarray, true_mask: np.ndarray) -> float:
        """Calculate pixel accuracy"""
        correct = (pred_mask == true_mask).sum()
        total = pred_mask.size
        return correct / total

    def evaluate(self, test_images: List[np.ndarray],
                test_masks: List[np.ndarray]) -> Dict:
        """Comprehensive model evaluation"""
        ious, dices, accuracies = [], [], []

        for img, true_mask in tqdm(zip(test_images, test_masks),
                                  desc="Evaluating", total=len(test_images)):
            # Get prediction
            raw_mask = self.model.predict_mask(img)
            clean_mask = self.model.postprocess_mask(raw_mask)

            # Calculate metrics
            iou = self.calculate_iou(clean_mask, true_mask)
            dice = self.calculate_dice(clean_mask, true_mask)
            accuracy = self.calculate_accuracy(clean_mask, true_mask)

            ious.append(iou)
            dices.append(dice)
            accuracies.append(accuracy)

        return {
            'mean_iou': float(np.mean(ious)),
            'mean_dice': float(np.mean(dices)),
            'mean_accuracy': float(np.mean(accuracies)),
            'std_iou': float(np.std(ious)),
            'std_dice': float(np.std(dices)),
            'std_accuracy': float(np.std(accuracies)),
            'samples_evaluated': len(test_images)
        }

# ==================== Visualization Utilities ====================
class VisualizationUtils:
    @staticmethod
    def plot_comparison(original: np.ndarray, result: np.ndarray,
                       mask: np.ndarray, title: str = "Segmentation Result"):
        """Plot original image, mask, and result side by side"""
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))

        # Original image
        axes[0].imshow(original)
        axes[0].set_title("Original Image")
        axes[0].axis('off')

        # Prediction mask
        axes[1].imshow(mask, cmap='gray')
        axes[1].set_title("Predicted Mask")
        axes[1].axis('off')

        # Result
        axes[2].imshow(result)
        axes[2].set_title("Extracted Object")
        axes[2].axis('off')

        plt.suptitle(title, fontsize=16, fontweight='bold')
        plt.tight_layout()
        return fig

    @staticmethod
    def result_to_bytes(result_image: np.ndarray, format: str = 'PNG') -> bytes:
        """Convert result image to bytes"""
        img_pil = Image.fromarray(result_image)
        img_byte_arr = io.BytesIO()
        img_pil.save(img_byte_arr, format=format)
        return img_byte_arr.getvalue()

# ==================== FastAPI Backend Integration ====================
from fastapi import FastAPI, File, UploadFile, HTTPException
from fastapi.responses import Response
import uvicorn

app = FastAPI(title="Segmentation Model API", version="1.0.0")

# Global model instance
segmenter = SegmentationModel("deeplabv3_segmentation.pth")

@app.get("/")
async def root():
    return {"message": "Segmentation Model API", "status": "active"}

@app.get("/health")
async def health_check():
    return {"status": "healthy", "device": str(device)}

@app.post("/segment")
async def segment_image(
    file: UploadFile = File(...),
    background: str = "black",
    use_tta: bool = True
):
    """Segment image and return result with specified background"""
    if not file.content_type.startswith('image/'):
        raise HTTPException(400, "File must be an image")

    # Map background string to RGB tuple
    bg_colors = {
        "black": (0, 0, 0),
        "white": (255, 255, 255),
        "green": (0, 255, 0),
        "blue": (0, 0, 255)
    }
    background_color = bg_colors.get(background, (0, 0, 0))

    try:
        image_data = await file.read()
        result = segmenter.process_single_image(image_data, background_color, use_tta)

        if not result['success']:
            raise HTTPException(500, f"Processing failed: {result.get('error', 'Unknown error')}")

        # Convert result to bytes
        result_bytes = VisualizationUtils.result_to_bytes(result['result_image'])

        return Response(
            content=result_bytes,
            media_type="image/png",
            headers={
                "X-Foreground-Ratio": f"{result['foreground_ratio']:.4f}",
                "X-Original-Size": f"{result['original_shape'][1]}x{result['original_shape'][0]}"
            }
        )

    except Exception as e:
        raise HTTPException(500, f"Processing error: {str(e)}")

@app.post("/segment/json")
async def segment_image_json(file: UploadFile = File(...)):
    """Segment image and return JSON with metrics"""
    if not file.content_type.startswith('image/'):
        raise HTTPException(400, "File must be an image")

    try:
        image_data = await file.read()
        result = segmenter.process_single_image(image_data)

        if result['success']:
            return {
                "success": True,
                "foreground_ratio": result['foreground_ratio'],
                "original_dimensions": {
                    "width": result['original_shape'][1],
                    "height": result['original_shape'][0]
                },
                "processing_time": "N/A"
            }
        else:
            return {
                "success": False,
                "error": result.get('error', 'Unknown error')
            }

    except Exception as e:
        return {
            "success": False,
            "error": str(e)
        }

# ==================== Main Execution ====================
if __name__ == "__main__":
    # Demonstration
    print("🚀 Starting Segmentation Model Backend")

    # Initialize model
    model = SegmentationModel("deeplabv3_segmentation.pth")

    # Example usage
    print("✅ Backend ready!")
    print("📝 To run API server: uvicorn backend:app --host 0.0.0.0 --port 8000")

    # Run API server if executed directly
    uvicorn.run(app, host="0.0.0.0", port=8000)
```
